```python
# Treat missing values using contextual median imputation

# Rainfall: fill missing values using the median of the same season
df["Rainfall_mm"] = df.groupby("Season")["Rainfall_mm"].transform(
    lambda x: x.fillna(x.median())
)

# Soil moisture: fill missing values using the median of the same season
df["Soil_Moisture_pct"] = df.groupby("Season")["Soil_Moisture_pct"].transform(
    lambda x: x.fillna(x.median())
)

# Yield: fill missing values using the median of the same crop and season
df["Yield_Tonnes_Ha"] = df.groupby(["Crop", "Season"])["Yield_Tonnes_Ha"].transform(
    lambda x: x.fillna(x.median())
)

print("Missing values treated using contextual median imputation.")
```

### 7.11 Outlier Analysis

Outlier analysis was performed to identify unusually high or low observations in important agricultural variables.

The Interquartile Range (IQR) method was used to identify potential outliers. The IQR is calculated as the difference between the third quartile (Q3) and the first quartile (Q1). Observations below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are considered potential outliers.

Outliers were identified for yield, production, profit, water usage, and water efficiency. These observations were not removed automatically because extreme values may represent genuine differences between farms.

```python
# Outlier Analysis using the IQR method

outlier_variables = [
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Profit_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3"
]

outlier_summary = []

for variable in outlier_variables:
    Q1 = df[variable].quantile(0.25)
    Q3 = df[variable].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[variable] < lower_bound) |
        (df[variable] > upper_bound)
    ]

    outlier_summary.append({
        "Variable": variable,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier Percentage": (len(outliers) / len(df)) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.round(2)
```

### 7.12 Outlier Summary Visualization

The number of potential outliers identified using the IQR method was visualized for the selected agricultural variables.

The comparison helps identify which variables contain a relatively higher number of extreme observations and supports further interpretation of the dataset's variability.

```python
# Visualize the number of potential outliers

plt.figure(figsize=(10, 6))

bars = plt.bar(
    outlier_summary["Variable"],
    outlier_summary["Outlier Count"]
)

plt.title(
    "Number of Potential Outliers by Variable",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Variable", fontsize=12)
plt.ylabel("Number of Potential Outliers", fontsize=12)

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)

# Add values above bars
for bar, value in zip(bars, outlier_summary["Outlier Count"]):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value}",
        ha="center",
        va="bottom",
        fontsize=11,
        fontweight="bold"
    )

plt.tight_layout()
plt.show()
```

**Observation:**  
Crop yield has a large difference between its mean and median, indicating that the distribution is strongly influenced by higher-value observations.

**Interpretation:**  
Most farm records have relatively lower yield values, while a smaller number of records have substantially higher yields. This creates a right-skewed distribution and explains why the mean is considerably higher than the median.

**Limitation:**  
The analysis describes the distribution but does not determine whether the extreme yield values are errors or genuine agricultural observations. They should therefore not be removed without additional validation.

Most farm records have relatively lower production values, while a smaller number of records show much higher production. This creates a right-skewed distribution and results in the mean being considerably higher than the median.

**Limitation:**  
The analysis does not determine whether high-production observations are unusual errors or genuine farm records. The observations should therefore be investigated before any decision is made to remove them.

**Limitation:**  
An IQR-based outlier is a statistical classification and does not necessarily indicate an incorrect observation. Therefore, these observations were retained rather than removed automatically.
